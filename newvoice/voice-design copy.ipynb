```python
import os
os.environ["CUDA_VISIBLE_DEVICES"]="0"
```

```python
from omnivoice import OmniVoice
import soundfile as sf
import torch
```

```python
from IPython.display import Audio
```

```python
model = OmniVoice.from_pretrained(
    "k2-fsa/OmniVoice",
    device_map="cuda",
    dtype=torch.float16
)
# Apple Silicon users: use device_map="mps" instead
```

```python
thai_text ="สวัสดีครับ เรากำลังทดสอบระบบแปลงข้อความเป็นเสียงภาษาไทย เพื่อตรวจสอบความชัดเจนของการออกเสียงวรรณยุกต์ และจังหวะการพูดที่ดูเป็นธรรมชาติที่สุด"
# 3. Generate cute anime voice using Voice Design
audios = model.generate(
    text=thai_text,
    language="Thai",  # ensures proper Thai tonal synthesis
    instruct="female, teenager, high pitch",
    speed=1.05,
)
sf.write("out3.wav", audios[0], 24000)
Audio("out3.wav")
```

```python
thai_text ="สวัสดีครับ เรากำลังทดสอบระบบแปลงข้อความเป็นเสียงภาษาไทย เพื่อตรวจสอบความชัดเจนของการออกเสียงวรรณยุกต์ และจังหวะการพูดที่ดูเป็นธรรมชาติที่สุด"
# 3. Generate cute anime voice using Voice Design
audios = model.generate(
    text=thai_text,
    language="Thai",  # ensures proper Thai tonal synthesis
    instruct="female, teenager, high pitch, japanese accent",
    speed=1.05,
)
sf.write("out4.wav", audios[0], 24000)
Audio("out4.wav")
```
